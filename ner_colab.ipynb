## 1. Cargar modelo

```python
nlp = spacy.load("es_core_news_sm")
print("Modelo cargado")
```

## 2. Textos de ejemplo

```python
textos = [
    "Lionel Messi juega en el Inter Miami en Estados Unidos.",
    "Apple presentó el nuevo iPhone en California.",
    "El presidente de Argentina visitó Madrid para reunirse con el rey.",
    "Google abrió una oficina en Buenos Aires.",
    "Cristiano Ronaldo fichó por Al-Nassr en Arabia Saudita."
]
print(len(textos), "textos")
```

## 3. Extraer entidades

```python
def extraer_entidades(texto):
    doc = nlp(texto)
    entidades = [(ent.text, ent.label_) for ent in doc.ents]
    return entidades

resultados = []
for texto in textos:
    ents = extraer_entidades(texto)
    resultados.append({"texto" : texto, "entidades" : ents})
    print(f"Texto: {texto}")
    print(f"Entidades: {ents}\n")

df = pd.DataFrame(resultados)
df.head()
```
